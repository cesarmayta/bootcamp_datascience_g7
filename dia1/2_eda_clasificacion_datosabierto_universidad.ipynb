# CARGAMOS DATASET

```python
PATH_CSV = '/content/drive/MyDrive/CODIGOG7/MODULO9/2-EDA/Datos_abiertos_admision_2021_1_2024_1.csv'
```

```python
df = pd.read_csv(PATH_CSV)
df
```

# EDA

```python
df_cleaned = df[(df['MODALIDAD'] == 'ORDINARIO') & (df['CICLO_POSTULA'] == 1)]
df_cleaned
```

```python
df_cleaned = df_cleaned.drop(columns=['IDHASH', 'COLEGIO', 'COLEGIO_PROV','COLEGIO_DIST','COLEGIO_PAIS','DOMICILIO_DEPA','DOMICILIO_PROV','DOMICILIO_DIST','NACIMIENTO_PAIS','NACIMIENTO_DEPA','NACIMIENTO_PROV','NACIMIENTO_DIST','CALIF_FINAL','MODALIDAD','CICLO_POSTULA'])
df_cleaned
```

```python
df_cleaned['egreso_postulacion'] = df_cleaned['ANIO_POSTULA'] - df_cleaned['COLEGIO_ANIO_EGRESO']
df_cleaned['edad'] = df_cleaned['ANIO_POSTULA'] - df_cleaned['ANIO_NACIMIENTO']
df_cleaned
```

```python
df_cleaned = df_cleaned.drop(columns=['COLEGIO_ANIO_EGRESO', 'ANIO_POSTULA','ANIO_NACIMIENTO'])
df_cleaned
```
