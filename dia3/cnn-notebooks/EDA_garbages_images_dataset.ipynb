```python
BASE_DIR = '/root/.cache/kagglehub/datasets/zlatan599/garbage-dataset-classification/versions/5/Garbage_Dataset_Classification/images'
```

```python
import os
import random
from PIL import Image
import shutil

# Define the new dataset directory, image size, and number of samples
NEW_DATASET_DIR = 'garbage_dataset'
IMAGE_SIZE = (64, 64)
NUM_SAMPLES_PER_SUBSET = 10
SUBSETS = ['train', 'validation', 'test']

print(f"Starting image processing from: {BASE_DIR}")
print(f"Creating new dataset directory: {NEW_DATASET_DIR}")
print(f"Resizing images to: {IMAGE_SIZE}")
print(f"Sampling {NUM_SAMPLES_PER_SUBSET} unique images per category for each subset ({', '.join(SUBSETS)})")

# 1. Create the main dataset directory. Remove if it already exists for a clean run.
if os.path.exists(NEW_DATASET_DIR):
    print(f"Removing existing directory: {NEW_DATASET_DIR}")
    shutil.rmtree(NEW_DATASET_DIR)
os.makedirs(NEW_DATASET_DIR)
print(f"Created {NEW_DATASET_DIR}")

# 2. Get the list of categories (subdirectories) from the BASE_DIR
categories = [d for d in os.listdir(BASE_DIR) if os.path.isdir(os.path.join(BASE_DIR, d))]
print(f"Found categories: {categories}")

# 3. Process each category
for category in categories:
    category_src_path = os.path.join(BASE_DIR, category)
    # Filter for common image file extensions
    image_files = [f for f in os.listdir(category_src_path) if f.lower().endswith(('.png', '.jpg', '.jpeg', '.gif', '.bmp'))]
    random.shuffle(image_files) # Shuffle images once for this category to ensure random unique selection

    current_image_index = 0
    print(f"\nProcessing category: {category} (found {len(image_files)} images)")

    # Iterate through each subset (train, validation, test)
    for subset in SUBSETS:
        subset_dest_path = os.path.join(NEW_DATASET_DIR, subset, category)
        os.makedirs(subset_dest_path, exist_ok=True)
        print(f"  Creating subset directory: {subset_dest_path}")

        samples_taken_for_subset = 0
        # Copy and process images until NUM_SAMPLES_PER_SUBSET is reached or no more images are available
        while samples_taken_for_subset < NUM_SAMPLES_PER_SUBSET and current_image_index < len(image_files):
            src_image_name = image_files[current_image_index]
            src_image_path = os.path.join(category_src_path, src_image_name)
            dest_image_path = os.path.join(subset_dest_path, src_image_name)

            try:
                img = Image.open(src_image_path)
                img = img.resize(IMAGE_SIZE) # Resize the image
                img.save(dest_image_path) # Save the processed image
                samples_taken_for_subset += 1
            except Exception as e:
                print(f"    Error processing image {src_image_path}: {e}")

            current_image_index += 1 # Move to the next unique image for the next split

        print(f"  Copied {samples_taken_for_subset} images to {subset}/{category}")
        if samples_taken_for_subset < NUM_SAMPLES_PER_SUBSET:
            print(f"  Warning: Only {samples_taken_for_subset} images were available for {category} in {subset}.")

print(f"\nDataset creation and processing complete in {NEW_DATASET_DIR}")
```
